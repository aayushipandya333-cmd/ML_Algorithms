```python
dataset = pd.DataFrame(house_data.data)      # Converting data into a dataframe using pandas and taking it in variable named dataset
```

```python
dataset
```

```python
dataset.columns = house_data.feature_names      # changing column haeders of dataset from numbers to actual headings 
```

```python
dataset                                  # Print entire dataset
```

```python
x=dataset                                # Independent features
y=house_data.target                      # Dependent feature
```

```python
x
```

```python
y
```

```python
from sklearn.model_selection import train_test_split                        # Train test split

# Split into 80% training and 20% testing
x_train, x_test, y_train, y_test = train_test_split(
    x, y, test_size=0.2, random_state=42
)
```

```python
x_train
```

```python
x_test
```

```python
from sklearn.preprocessing import StandardScaler
scaler = StandardScaler()
```

```python
x_train = scaler.fit_transform(x_train)      # scaler.fit_transform(x_train) does two things in one step:

                                             # fit() - Calculates the statistical parameters (like mean and standard deviation) from your training data

                                             # transform() - Applies those parameters to scale/transform the training data
```

```python
x_test = scaler.transform(x_test)             # fit_transform() on training data = learns patterns from training

                                              # transform() on test data = applies the SAME patterns to test data
                                             # NEVER fit() on test data - this would cause data leakage!
                                             # This would make your model perform well on test data but fail in production because the test data statistics won't match real-world data.
```

We can work with actual values, but many machine learning algorithms perform poorly or incorrectly if we don't scale the data. Transformations help algorithms treat all features fairly, regardless of their original units or magnitudes.

Imagine you have a dataset with these features:

**Feature  	Values	                 Unit**
Size	    500 - 5,000	             sq ft
Bedrooms	1 - 5	                 count
Price	    $50,000 - $1,000,000  	dollars

1. The Magnitude Problem (Distance-Based Algorithms)
Algorithms like K-Nearest Neighbors, K-Means, SVM, PCA use distance calculations (Euclidean distance).
2. The Gradient Descent Problem (Gradient-Based Algorithms)
Algorithms like Linear Regression, Logistic Regression, Neural Networks use gradient descent to find optimal weights.  Gradient descent takes a zigzag path to reach minimum
3. The Regularization Problem
Algorithms with regularization (L1/L2) penalize large coefficients.
4. The Interpretability Problem
Without scaling: Coefficients are in different units, making them hard to compare.

```python
from sklearn.linear_model import LinearRegression

# Cross validation
from sklearn.model_selection import cross_val_score    
# cross_val_score : function that performs cross-validation.  
# A resampling statistical technique used to evaluate how well a machine learning model generalizes to unseen data. 
# By systematically partitioning a dataset into multiple subsets, it ensures that every data point is used for both training and validation. 
```

```python
regression = LinearRegression()
regression.fit(x_train,y_train)
```

```python
mse=cross_val_score (regression, x_train, y_train, scoring='neg_mean_squared_error', cv=5)
```

cross_val_score(...)	            Function that trains & tests the model multiple times,  
regression	                        The model to evaluate,  
x_train, y_train                	Training data (features & target),  
scoring='neg_mean_squared_error'	Use Mean Squared Error (MSE) as the evaluation metric
(Note: It returns negative MSE because sklearn maximizes scores, so smaller MSE = larger negative number),  
cv=5	                            5-fold cross-validation – splits data into 5 parts, trains on 4, tests on 1, repeats 5 times
