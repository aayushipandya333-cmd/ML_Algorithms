```python
def plotting(var,num):
    plt.subplot(2,2,num)
    sns.histplot(data[var], kde = True)

plotting('Age',1)
plotting('RestingBP',2)
plotting('Cholesterol',3)
plotting('MaxHR',4)

plt.tight_layout()
```

#### Analysis  : Here Age and MaxHR is normally distributed, but in RestingBP we can see that it is towards right side, because 1 person has resting BP 0, which is not possible. So it may be a wrong entry. We can delete it.
#### Also in cholestrol, many people have colestrol 0 which is again not possible.

```python
data['Cholesterol'].value_counts()        # 172 people have 0 cholestrol --> Not possible
```

```python
ch_mean = data.loc[data['Cholesterol']!= 0 , 'Cholesterol'].mean()
ch_mean
```

```python
data['Cholesterol'] = data['Cholesterol'].replace(0, ch_mean)           # 'Cholesterol' with value 0 is replaced by mean value
data['Cholesterol'] = data['Cholesterol'].round(2)
ch_mean = data.loc[data['Cholesterol']!= 0 , 'Cholesterol'].mean()
ch_mean
```

```python
bp_mean = data.loc[data['RestingBP']!= 0 , 'RestingBP'].mean()
bp_mean
```

```python
data['RestingBP'] = data['RestingBP'].replace(0, bp_mean)           # 'Cholesterol' with value 0 is replaced by mean value
data['RestingBP'] = data['RestingBP'].round(2)
bp_mean = data.loc[data['RestingBP']!= 0 , 'RestingBP'].mean()
bp_mean
```
