# Feature Extraction/ Selection 

#### Pearson's correlation coefficient (r) measures the strength and direction of the linear relationship between two continuous variables

```python
from scipy.stats import pearsonr

selected_features = ['age', 'is_female', 'bmi', 'children', 'is_smoker',
       'region_northeast', 'region_northwest', 'region_southeast',
       'region_southwest', 'bmi_category_Underweight', 'bmi_category_Normal',
       'bmi_category_Overweight', 'bmi_category_Obese']

correlations = {
    feature : pearsonr(data_cleaned[feature], data_cleaned['charges'])[0]      # pearsonr() returns two values, [0] → correlation, [1] → p-value, here we want only correlation
    for feature in selected_features
}

correlation_data = pd.DataFrame(list(correlations.items()), columns = ['Feature', 'Pearson Correlation'])
correlation_data.sort_values(by='Pearson Correlation', ascending = False)
```

```python
# We can find is_smoker is highly correlated
```

#### Chi-Square Test (χ²) 

#### The Chi-Square test is a statistical test mainly used to determine whether two categorical variables are related/associated with each other.

#### For example, in an insurance dataset:

#### smoker → categorical (yes/no)
#### region → categorical (north/south/east/west)

#### You can use a Chi-Square test to ask:

#### Is smoking status associated with region?

```python
categorical_features = ['is_female','is_smoker',
       'region_northeast', 'region_northwest', 'region_southeast',
       'region_southwest', 'bmi_category_Underweight', 'bmi_category_Normal',
       'bmi_category_Overweight', 'bmi_category_Obese']
```

```python
from scipy.stats import chi2_contingency

alpha = 0.05

data_cleaned['charges_bin'] = pd.qcut(data_cleaned['charges'], q=4, labels=False)
chi2_results = {}

for col in categorical_features :
    contingency = pd.crosstab(data_cleaned[col], data_cleaned['charges_bin'])
    chi2_stat, p_val, _, _ = chi2_contingency(contingency)
    decision = "Reject Null (Keep Feature)" if p_val < alpha else "Accept Null (Drop Feature)"
    chi2_results[col] = {
    'chi2_statistic' : chi2_stat,
    'p_value' : p_val,
    'Decision'  : decision
    }


chi2_data = pd.DataFrame(chi2_results).T
chi2_data = chi2_data.sort_values(by='p_value')
chi2_data
    
```
